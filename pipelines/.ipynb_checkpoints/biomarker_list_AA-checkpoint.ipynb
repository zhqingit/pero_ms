## Description

This notebook does the differential expression analysis based on the integrated AA and GTEx data (normal)

[GSE154548](https://www.ncbi.nlm.nih.gov/bioproject/PRJNA646641) and [GSE76987](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE76987) 
## Method
We use the same method to that in the [CRC biomarker selection](https://github.com/GeneGenieDx/ggdx/blob/main/projects/crc/Gene_screen/biomarker_list_generation_TCGA_GTEx_DE_analysis.ipynb) except removing the batch correction step, because the two AA datasets have no attached normal samples. But we prove the batch effect is very small in this analysis.

### 1.6 combine gtex and AA dataset, as well as TCGA
#### 1.6.1 combine gtex and AA dataset, which is used in the DE analysis

```r
genes <- intersect(rownames(counts_aa),rownames(colon.gtex))
counts_raw <- cbind(counts_aa[genes,],colon.gtex[genes,])
dim(counts_raw)
head(counts_raw)
#create the meta information
sources <- c(source_aa,rep("GTEX",ncol(colon.gtex)))
labels <- c(rep("AA",ncol(counts_aa)),rep("normal",ncol(colon.gtex)))
metas<-cbind(sources, labels)
table(metas[,"labels"])
```

## 2. batch effect detection
### 2.1 batch effect detection

```r
# Group annoation for AA, CRC and normal samples
Group <- factor(labels_with_tcga)
y <- DGEList(counts=as.matrix(counts_raw_tcga),group=Group)
#colnames(y) <- colnames(tcga_gtex)
dim(y)
y$samples$lib.size <- colSums(y$counts)
y <- calcNormFactors(y)
cols <- sources_with_tcga
cols[grep("gse154548",cols)] <- "orange"
cols[grep("gse76987",cols)] <- "purple"
cols[grep("GTEX",cols)] <- "blue"
cols[grep("TCGA",cols)] <- "red"
# MDS plot to check similarity between samples
plotMDS(y, labels = Group, col=cols)
```

## 3. Differential expressed genes
### 3.1 differential analysis

```r
# Group annoation for AA and normal samples
Group <- factor(metas[,"labels"])
y1 <- DGEList(counts=as.matrix(counts_raw),group=Group)

y1$samples$lib.size <- colSums(y1$counts)
y1 <- calcNormFactors(y1)
```

```r
design <- model.matrix(~0+Group)
colnames(design)

# Estimate dispersion
x <- estimateGLMTrendedDisp(y1, design)
fit <- glmFit(x,design,robust = TRUE)
# Likelyhood ratio DE test
lrt <- glmLRT(fit, contrast = c(1,-1))
top <- topTags(lrt, n=Inf)
# Summary the DE results
summary(de <- decideTestsDGE(lrt))
result <- as.data.frame(top$table)
```

```r
# Add reads count to the DE results
count_de <- counts_raw[rownames(result) ,]

#get a vector of labels
disease_labels=metas[,"labels"]
ndisease <- table(disease_labels)["AA"]
ndisease

# Add the percentage of Tumor/nomral sample expressed for each gene to the DE results
result["T_perc"] <- rowSums(cpm(count_de[,1:ndisease])>1)/ndisease
result["N_perc"] <- rowSums(cpm(count_de[,(ndisease+1):dim(count_de)[2]])>1)/(dim(count_de)[2] -ndisease)

##Add median tumor and normal median
result["T_median"] <- apply(cpm(count_de[,1:ndisease]), 1, median)
result["N_median"] <- apply(cpm(count_de[,(ndisease+1):dim(count_de)[2]]), 1, median)
result["AUC"]<-apply(count_de, 1, auc, response=disease_labels, quiet=TRUE)



# Add metric fold change to the DE results, which is log2FC * (T_perc + 0.001)/(N_perc + 0.001)
result["Metric_FC"]<- result$logFC * (result["T_perc"] + 0.001)/(result["N_perc"] + 0.001)
result["Metric_FC_CPM"]<- result$Metric_FC*result$logCPM 

head(result)


```

```r
# read ensembl id to gene name
id2sym <- read.csv(file="data_TCGA_GTEX/IDToGene.csv",header=T,stringsAsFactors=F)
head(id2sym)
```

```r
##adding the gene names, normal percentile and tumor percentile
#normal percentile calculation
final_results=result[order(result$N_median, decreasing=TRUE), ]
final_results["N_percentile"]=1-(seq(1, nrow(final_results), by=1)/nrow(final_results))

#tumor percentile calculation
final_results=final_results[order(final_results$T_median, decreasing=TRUE), ]
final_results["T_percentile"]=1-(seq(1, nrow(final_results), by=1)/nrow(final_results))

Gene <- id2sym[match(rownames(final_results),id2sym[,1]),2] 
final_results <- cbind(Gene,final_results)

head(final_results)
```

```r
write.table(final_results,file="./Results/GeneLists/AA_GeneList.tsv",sep = "\t",quote = FALSE, row.names=T)
```

```r
############### List of Genes 1 Select those genes with an AUC >0.9 with a gene expression in normal
############### tissue < 0.12 percentile so low expression in most normal samples sort this list
############### by the median tumor expression from highest to lowest.
lowBackgroundIndex=which(final_results$FDR<0.001 & final_results$AUC>0.9 & 
                         final_results$N_percentile<0.12 & final_results$logFC>2)

lowBackground_Results=final_results[lowBackgroundIndex, ]
print(c("Number of promising genes with low expression in normal samples", nrow(lowBackground_Results)))

lowBackground_Results=lowBackground_Results[order(lowBackground_Results$T_median, decreasing=TRUE), ]



## write results to a file
write.table(lowBackground_Results,file="./Results/GeneLists/lowBackground_GeneList.tsv",sep = "\t",quote = FALSE, row.names=F)
head(lowBackground_Results)

```

```r
############### List of Genes 1 Select those genes with an AUC >0.9 and a FDR <0.001
###############  and a logFC of >2 the logFC cutoff  is to ensure the expression in tumor is higher than the expression in normal
###############sort this list by the median tumor expression from highest to lowest.

######This list is genes that we don't care very much about the background level of expression in normal just
######that the expression in cancer be very high 

highAAIndex=which(final_results$FDR<0.001 & final_results$AUC>0.9 & final_results$logFC>2)

highAA_Results=final_results[highAAIndex, ]
print(c("Number of promising genes with very high expression in AA samples", nrow(highAA_Results)))

highAA_Results=highAA_Results[order(highAA_Results$T_median, decreasing=TRUE), ]

head(highAA_Results)
#highAA_Results[1:10, c("FDR", "T_perc", "logFC", "N_perc", "T_median", "N_median", "AUC", "Gene", "N_percentile", "T_percentile")]
dim(highAA_Results)


## write results to a file
write.table(highAA_Results,file="./Results/GeneLists/highAA_GeneList.tsv",sep = "\t",quote = FALSE, row.names=F)
```

## 5. Generate gene groups for primer design experiments
## the goal is to generate a list of gene groups of size n (=8 by default) that contains all genes from sorted highAA and lowBG list
### Algorithm
#### step a: create a new empty group A 
##### definition:
######    n_hc: the number of genes from highAA list already in A
######    n_bg: the number of genes from lowBG list already in A
######    n_hc_pick: how many highAA genes will be added to A
######    n_bg_pick: how many lowBG genes will be added to A
##### process:
    
###### 1. calculate how many genes (n_hc_pick, n_bg_pick) to extract from sorted highAA and lowBG lists respectively
######         if A is empty, n_hc_pick = n_bg_pick = n/2 else caculate n_hc_pick and n_bg_pick based on the n_hc and n_bg to make 
######         the numbers of genes from highAA and lowBG lists in A are equal as much as possible
###### 2. extract the top n_hc_pick and n_bg_pick genes from the two lists respectively, and add them to A
###### 3. remove the duplicates
###### 4. remove the genes in A from highAA and lowBG lists
###### 5. repeat from step 1 till the total gene number in A is equal to n
#### repeat step "a" till both highAA and lowBG lists are empty

```r
highAA_gene = highAA_Results[,"Gene"]
lowBG_gene = lowBackground_Results[,"Gene"]
cat("total genes:",length(unique(c(highAA_gene,lowBG_gene))),"\n")
A <- c()
gg <- list()
n <-8
while (length(highAA_gene)>0 | length(lowBG_gene)>0){
    n_pick <- n - length(A) 
    if (n_pick == 0){
        gg[[length(gg)+1]] = A
        A <- c()
        n_pick <- n - length(A) 
    }
    n_hc = length(!is.na(match(A,highAA_gene)))
    n_bg = length(!is.na(match(A,lowBG_gene)))
    if ((n_hc+n_bg)>0){
        n_hc_pick = as.integer(n_pick * n_bg/(n_hc+n_bg))
        n_bg_pick = n_pick -n_hc_pick
    }else{
        n_hc_pick = as.integer(n/2)
        n_bg_pick = n - n_hc_pick
    }
    
    
    if (length(highAA_gene)>0){
        if (n_hc_pick >0) {
            if (n_hc_pick  > length(highAA_gene)){
                A = c(A, highAA_gene[1:length(highAA_gene)])
            }else{
                A = c(A, highAA_gene[1:n_hc_pick])
            }
        }
    }
    
    
    if (length(lowBG_gene)> 0){
        if (n_bg_pick >0){
            if (n_bg_pick > length(lowBG_gene)){
                A = c(A, lowBG_gene[1:length(lowBG_gene)])
            }else{
                A = c(A, lowBG_gene[1:n_bg_pick])
            }
        }
    }
    A<- unique(A)
    if (length(lowBG_gene)>0){
        rem.ids.bg <- na.omit(match(A,lowBG_gene))
        if (length(rem.ids.bg) > 0) lowBG_gene <- lowBG_gene[-rem.ids.bg]
    }
    if (length(highAA_gene)>0){
        rem.ids.hc <- na.omit(match(A,highAA_gene))
        if (length(rem.ids.hc)>0) highAA_gene <- highAA_gene[-rem.ids.hc]
    }
}
if (length(A)>0) gg[[length(gg)+1]] <- A
```

```r
length(unlist(gg))
can_label_Results <- highAA_Results
group <- rep(0,nrow(can_label_Results))
for (i in c(1:length(gg))){
    ids <- match(gg[[i]],can_label_Results[,"Gene"])
    group[ids] <- i
}
can_label_Results_1 <- cbind(can_label_Results,group)
head(can_label_Results_1)

## write all candidate genes
write.table(can_label_Results_1,file="./Results/GeneLists/candidate_label_GeneList.tsv",sep = "\t",quote = FALSE, row.names=F)
```
