## Compare to qPCR

```r
TO <- function(mat,a,b){
    #filter the qPCR by titrition order
    #two filters:
    # 1. a*q_a >= b*q_c & a*q_c >= b*q_d & a*q_d >= b*q_b
    # 2. a*q_a <= b*q_c & a*q_c <= b*q_d & a*q_d <= b*q_b
    pass = apply(mat,1,function(x,a,b){
        if (a*x[1]>=b*x[3] & a*x[3]>=b*x[4] & a*x[4]>=b*x[2]) return(TRUE)
        if (b*x[1]<=a*x[3] & b*x[3]<=a*x[4] & b*x[4]<=a*x[2]) return(TRUE)
        return(FALSE)
    },a=a,b=b)
    return(mat[which(pass),])
}
```

```r
EMR <- function(mat,a,b){
    #filter the qPCR by mixture ratio
    z = 1.43
    
    pass = apply(mat,1,function(x,a,b){
        R_ab = x[1]/x[2]
        R_cd = x[3]/x[4]
        EMR_cd = (3*z*R_ab+1)/(z*R_ab+3)*(z+3)/(3*z+1)
        #R_ab_lower = b*R_ab
        #R_ab_upper = a*R_ab
        R_cd_lower = b*R_cd
        R_cd_upper = a*R_cd
        EMR_cd_lower = b*EMR_cd
        EMR_cd_upper = a*EMR_cd
        
        if (R_cd >= EMR_cd & R_cd_lower <= EMR_cd_upper) return(TRUE)
        if (R_cd <= EMR_cd & R_cd_upper >= EMR_cd_lower) return(TRUE)
        return(FALSE)
    },a=a,b=b)
    return(mat[which(pass),])
}
```

```r
# load the qPCR data 
qPCR = read.csv(file="cq_total.txt",header=T,row.names=1, stringsAsFactors=F,sep='\t')
```

```r
head(qPCR)
```

```r
#only keep the unique genes
dim(qPCR)
gid_uni = names(which(table(qPCR[,1])==1))
qPCR = qPCR[match(gid_uni,qPCR[,1]),]
rownames(qPCR) = qPCR[,1]
qPCR = qPCR[,-1]
dim(qPCR)
```

```r
# remove the instable genes based on the CT values' range
dim(qPCR)
ct.cf.min = 11
ct.cf.max = 32
qPCR = subset(qPCR,PCR_A >ct.cf.min & PCR_B > ct.cf.min & PCR_C >ct.cf.min & PCR_D >ct.cf.min & 
              PCR_A <= ct.cf.max & PCR_B <= ct.cf.max & PCR_C <=ct.cf.max & PCR_D <=ct.cf.max)
dim(qPCR)
```

```r
dim(qPCR)
qPCR = TO(qPCR, a=1.15, b=0.85)
qPCR = EMR(qPCR, a=1.15, b=0.85)
dim(qPCR)
```

```r
overlap_genes = intersect(rownames(qPCR),rownames(tpm))
tpm_overlap = tpm[overlap_genes,]
tpm_overlap = cbind(apply(tpm_overlap[,c(2,3)],1,mean),apply(tpm_overlap[,c(4,5)],1,mean))

colnames(tpm_overlap) = c("MAQCA_GSE47792_RNAseq","MAQCB_GSE47792_RNAseq")
#qPCR_overlap = qPCR[match(overlap_genes,rownames(qPCR)),c(1,2)]
qPCR_overlap = qPCR[overlap_genes,c(1,2)]
colnames(qPCR_overlap) = c("MAQCA_qPCR","MAQCB_qPCR")
dat_overlap = cbind(tpm_overlap, qPCR_overlap)
dim(dat_overlap)
```

```r
head(dat_overlap)
dim(dat_overlap)
cor(log2(dat_overlap))^2
cor(dat_overlap[,1],log2(dat_overlap[,3]))
```

```r
# estimation method from the paper
# the method is very sensitive to the low expression data, and it is
# not fare to compare the performance of our pipeline to the ones in the paper
# because the overall running enviroments are different
# so we print the values based on the method in the paper, but don't use it to 
# evaluate our pipeline. See the details about the paper in the below section "output in the paper"
median(abs(log2(dat_overlap[,1]/dat_overlap[,2]) - log2(dat_overlap[,3]/dat_overlap[,4])))
```

```r
dat_overlap_log =  dat_overlap
dat_overlap_log = log2(dat_overlap_log)
cor_rna_pcr = cor(dat_overlap)
cor_rna_pcr_log = cor(dat_overlap_log)^2
cor_rna_pcr
cor_rna_pcr_log
```

```r
## print the replicates from different experiments for MAQCB
#p1 = ggplot(data=dat_overlap_log, aes(x=MAQCA_GSE83402_RNAseq,y=-MAQCA_qPCR))+theme_classic()+geom_point(alpha=0.3)+
#annotate(geom="text",x=5,y=-3.8, label=paste("R^2=",round(cor_rna_pcr_log["MAQCA_GSE83402_RNAseq","MAQCA_qPCR"],2),sep=""))
#p2 = ggplot(data=dat_overlap_log, aes(x=MAQCB_GSE83402_RNAseq,y=-MAQCB_qPCR))+theme_classic()+geom_point(alpha=0.3)+
#annotate(geom="text",x=5,y=-3.8, label=paste("R^2=",round(cor_rna_pcr_log["MAQCB_GSE83402_RNAseq","MAQCB_qPCR"],2),sep=""))
p3 = ggplot(data=dat_overlap_log, aes(x=MAQCA_GSE47792_RNAseq,y=-MAQCA_qPCR))+theme_classic()+geom_point(alpha=0.3)+
annotate(geom="text",x=-1,y=-3.8, label=paste("R^2=",round(cor_rna_pcr_log["MAQCA_GSE47792_RNAseq","MAQCA_qPCR"],2),sep=""))+
annotate(geom="text",x=-1,y=-3.9, label=paste("R=",round(sqrt(cor_rna_pcr_log["MAQCA_GSE47792_RNAseq","MAQCA_qPCR"]),2),sep=""))
p4 = ggplot(data=dat_overlap_log, aes(x=MAQCB_GSE47792_RNAseq,y=-MAQCB_qPCR))+theme_classic()+geom_point(alpha=0.3)+
annotate(geom="text",x=-1,y=-3.6, label=paste("R^2=",round(cor_rna_pcr_log["MAQCB_GSE47792_RNAseq","MAQCB_qPCR"],2),sep=""))+
annotate(geom="text",x=-1,y=-3.7, label=paste("R=",round(sqrt(cor_rna_pcr_log["MAQCB_GSE47792_RNAseq","MAQCB_qPCR"]),2),sep=""))
```

### fold change (FC) on RNA-seq vs. qPCR

```r
dat_overlap_dif_log = log2(cbind(dat_overlap[,"MAQCB_GSE47792_RNAseq"]/dat_overlap[,"MAQCA_GSE47792_RNAseq"],
                            dat_overlap[,"MAQCB_qPCR"]/dat_overlap[,"MAQCA_qPCR"])
                      )

colnames(dat_overlap_dif_log) = c("MAQCB_MAQCA_GSE47792","MAQCB_MAQCA_qPCR")
cor_fc = cor(dat_overlap_dif_log)^2
dat_overlap_dif_log = as.data.frame(dat_overlap_dif_log,stringsAsFactors=F)
```

```r
## print the replicates from different experiments for MAQCB
#p1 = ggplot(data=dat_overlap_dif_log, aes(x=MAQCB_MAQCA_GSE83402,y=-MAQCB_MAQCA_qPCR))+theme_classic()+geom_point(alpha=0.3)+
#annotate(geom="text",x=5,y=1, label=paste("R^2=",round(cor_fc["MAQCB_MAQCA_GSE83402","MAQCB_MAQCA_qPCR"],2),sep=""))
p2 = ggplot(data=dat_overlap_dif_log, aes(x=MAQCB_MAQCA_GSE47792,y=-MAQCB_MAQCA_qPCR))+theme_classic()+geom_point(alpha=0.3)+
annotate(geom="text",x=-2,y=1, label=paste("R^2=",round(cor_fc["MAQCB_MAQCA_GSE47792","MAQCB_MAQCA_qPCR"],2),sep=""))+
annotate(geom="text",x=-2,y=0.8, label=paste("R=",round(sqrt(cor_fc["MAQCB_MAQCA_GSE47792","MAQCB_MAQCA_qPCR"]),2),sep=""))
```

```r
grid.arrange(p2,ncol=2,nrow=2,
             top=textGrob("FC on (RNAseq vs. qPCR) ",gp=gpar(fontsize=20,font=3)))
```
